```python
import matplotlib.pyplot as plt
import numpy as np


def build_overlay_exact_gray(
    img,
    results,
    T,
    m,
    grey_levels=(0.35, 0.65),
    period_box_alpha=0.05,   # 0.05 = 95% transparent
    draw_bbox=True,
    bbox_value=0.65,
):
    """
    Raster-exact overlay.

    Important:
    - period colour changes once per period
    - scan lines inside the same period use the same grey value
    - boxes are transparent
    - scan lines are exact pixel columns
    """
    base = np.asarray(img, dtype=float)

    if base.ndim == 3:
        base = base[..., 0]

    if base.max() > 1:
        base = base / base.max()

    H, W = base.shape
    out = base.copy()

    global_period_idx = 0

    for res in results:
        bbox = res["bbox"]
        x0, x1 = int(bbox["x0"]), int(bbox["x1"])
        y0, y1 = int(bbox["y0"]), int(bbox["y1"])
        line_id = bbox["line_id"]

        xs = list(range(x0, x1 + 1, m))

        samples = []
        for i, x in enumerate(xs):
            if i >= len(line_id):
                break
            if x < 0 or x >= W:
                continue

            col = base[y0:y1 + 1, x]
            ys = np.flatnonzero(col > 0)

            if len(ys) == 0:
                continue

            y_top = y0 + int(ys[0])
            y_bottom = y0 + int(ys[-1])
            lid = int(line_id[i])

            samples.append(
                {
                    "x": x,
                    "lid": lid,
                    "y_top": y_top,
                    "y_bottom": y_bottom,
                    "period_id": None,
                    "grey": None,
                }
            )

        # ------------------------------------------------------------
        # 1. Detect periods and assign one grey value per period
        # ------------------------------------------------------------
        period_start_i = None
        period_top = None
        period_bottom = None

        for sample_i, sample in enumerate(samples):
            x = sample["x"]
            lid = sample["lid"]
            y_top = sample["y_top"]
            y_bottom = sample["y_bottom"]

            if lid == 0:
                period_start_i = sample_i
                period_top = y_top
                period_bottom = y_bottom

            elif lid == T - 1 and period_start_i is not None:
                period_end_i = sample_i

                period_grey = grey_levels[global_period_idx % 2]

                x_left = samples[period_start_i]["x"]
                x_right = samples[period_end_i]["x"]

                x_left_clip = max(0, min(x_left, x_right))
                x_right_clip = min(W - 1, max(x_left, x_right))

                y_top_clip = max(0, min(period_top, period_bottom))
                y_bottom_clip = min(H - 1, max(period_top, period_bottom))

                # Transparent period box
                region = out[
                    y_top_clip:y_bottom_clip + 1,
                    x_left_clip:x_right_clip + 1
                ]

                out[
                    y_top_clip:y_bottom_clip + 1,
                    x_left_clip:x_right_clip + 1
                ] = (
                    (1 - period_box_alpha) * region
                    + period_box_alpha * period_grey
                )

                # Assign same grey to all scan lines in this period
                for j in range(period_start_i, period_end_i + 1):
                    samples[j]["period_id"] = global_period_idx
                    samples[j]["grey"] = period_grey

                global_period_idx += 1
                period_start_i = None
                period_top = None
                period_bottom = None

        # ------------------------------------------------------------
        # 2. Draw scan lines using period colour, not line-id colour
        # ------------------------------------------------------------
        for sample in samples:
            x = sample["x"]
            y_top = sample["y_top"]
            y_bottom = sample["y_bottom"]

            # Only draw lines that belong to a detected full period
            if sample["grey"] is None:
                continue

            g = sample["grey"]
            out[y_top:y_bottom + 1, x] = g

        # ------------------------------------------------------------
        # 3. Draw corridor bounding box directly into pixels
        # ------------------------------------------------------------
        if draw_bbox:
            xx0 = max(0, min(W - 1, x0))
            xx1 = max(0, min(W - 1, x1))
            yy0 = max(0, min(H - 1, y0))
            yy1 = max(0, min(H - 1, y1))

            out[yy0, xx0:xx1 + 1] = bbox_value
            out[yy1, xx0:xx1 + 1] = bbox_value
            out[yy0:yy1 + 1, xx0] = bbox_value
            out[yy0:yy1 + 1, xx1] = bbox_value

    return out
```
