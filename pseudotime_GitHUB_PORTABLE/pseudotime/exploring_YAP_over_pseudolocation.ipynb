```python
import ast
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.cluster import DBSCAN, AgglomerativeClustering
from sklearn.neighbors import NearestNeighbors
from sklearn.metrics import silhouette_score


# ============================================================
# 0. SETTINGS
# ============================================================

FEATURES = [1462, 1476]

min_profile_length = 6

checkpoints_dir = Path("checkpoints")
datasets_dir = Path("datasets")
out_dir = Path("clustering_outputs_both_surfaces")
out_dir.mkdir(exist_ok=True)

ks = [2, 3, 4, 5, 6]

n_pcs_cluster_default = 5
min_samples_default = 15
eps_values = [1.2, 1.4, 1.6, 1.8, 2.0]


# ============================================================
# 1. HELPER FUNCTIONS
# ============================================================

def parse_pseudotime_widths(x):
    """
    Convert pseudotime_widths into a numeric numpy array.
    Works for lists, numpy arrays, tuples, and stringified lists.
    """
    if isinstance(x, np.ndarray):
        return x.astype(float)

    if isinstance(x, list):
        return np.asarray(x, dtype=float)

    if isinstance(x, tuple):
        return np.asarray(list(x), dtype=float)

    if isinstance(x, str):
        try:
            parsed = ast.literal_eval(x)
            return np.asarray(parsed, dtype=float)
        except Exception:
            cleaned = (
                x.replace("[", "")
                 .replace("]", "")
                 .replace(",", " ")
                 .split()
            )
            return np.asarray(cleaned, dtype=float)

    raise ValueError(f"Cannot parse pseudotime_widths value: {x}")


def resample_ts_raw(ts, target_len):
    """
    Resample a variable-length confinement-width profile
    into a fixed-length profile.
    """
    ts = np.asarray(ts, dtype=float)

    if len(ts) == 1:
        return np.repeat(ts[0], target_len)

    x_old = np.linspace(0, 1, len(ts))
    x_new = np.linspace(0, 1, target_len)

    return np.interp(x_new, x_old, ts)


def prepare_df(df, prefix, keys):
    """
    Prefix CellProfiler columns to avoid collisions after merging.
    """
    rename_dict = {
        col: f"{prefix}_{col}"
        for col in df.columns
        if col not in keys
    }

    return df.rename(columns=rename_dict)


def show(obj):
    """
    Works in notebooks and scripts.
    """
    try:
        display(obj)
    except NameError:
        print(obj)


# ============================================================
# 2. MAIN FUNCTION FOR ONE SURFACE
# ============================================================

def run_surface_clustering(feature_idx):
    print("\n" + "=" * 100)
    print(f"PROCESSING FEATURE {feature_idx}")
    print("=" * 100)

    checkpoint_path = checkpoints_dir / f"pseudotime_cells_clean_{feature_idx}.pkl"
    outlier_path = datasets_dir / f"{feature_idx}_dapi_python_segmentation_outliers.txt"

    dapi_cp_path = datasets_dir / str(feature_idx) / "MyExpt_IdentifyPrimaryObjects.csv"
    phalloidin_cp_path = datasets_dir / str(feature_idx) / "MyExpt_Phalloidin_segmented.csv"
    yap_cp_path = datasets_dir / str(feature_idx) / "MyExpt_YAP_segmented.csv"

    # ============================================================
    # 2.1 Load pseudotime database
    # ============================================================

    SD_raw = pd.read_pickle(checkpoint_path)

    if "Feature_Idx" not in SD_raw.columns:
        SD_raw["Feature_Idx"] = feature_idx

    print("Loaded pseudotime database:", checkpoint_path)
    print("Initial SD_raw shape:", SD_raw.shape)

    # ============================================================
    # 2.2 Filter by pseudotime_widths length
    # ============================================================

    SD_clean = SD_raw.copy()

    SD_clean["pseudotime_widths"] = SD_clean["pseudotime_widths"].apply(parse_pseudotime_widths)
    SD_clean["pseudotime_widths_length"] = SD_clean["pseudotime_widths"].apply(len)

    n_initial = len(SD_clean)

    SD_clean = SD_clean[
        SD_clean["pseudotime_widths_length"] >= min_profile_length
    ].copy()

    n_after_length_filter = len(SD_clean)

    print(f"After pseudotime_widths length >= {min_profile_length}: {SD_clean.shape}")

    # ============================================================
    # 2.3 Remove manual DAPI segmentation outliers
    # ============================================================

    outliers = []

    if outlier_path.exists():
        with open(outlier_path, "r") as f:
            for line in f:
                name = line.strip()

                if name == "":
                    continue

                if not name.endswith(".tif"):
                    name = name + ".tif"

                outliers.append(name)

        SD_clean["dapi_filename"] = SD_clean["path_dapi"].apply(lambda p: Path(p).name)

        SD_clean = SD_clean[
            ~SD_clean["dapi_filename"].isin(outliers)
        ].copy()

    else:
        print("No outlier file found:", outlier_path)
        SD_clean["dapi_filename"] = SD_clean["path_dapi"].apply(lambda p: Path(p).name)

    n_after_outlier_filter = len(SD_clean)

    print(f"Manual outliers listed: {len(outliers)}")
    print(f"After manual outlier removal: {SD_clean.shape}")

    # ============================================================
    # 2.4 Prepare ImageNumber / ObjectNumber for CellProfiler merge
    # ============================================================

    # Important:
    # Do not reset index before this.
    # The original index encodes ImageNumber and ObjectNumber.
    SD_clean["ImageNumber"] = SD_clean.index // 10000
    SD_clean["ObjectNumber"] = SD_clean.index % 10000

    keys = ["Metadata_ObjNumber", "ObjectNumber"]

    cluster_map = SD_clean.rename(
        columns={"ImageNumber": "Metadata_ObjNumber"}
    )

    print("\nPrepared merge keys:")
    print(cluster_map[["Metadata_ObjNumber", "ObjectNumber"]].head())

    # ============================================================
    # 2.5 Load CellProfiler tables
    # ============================================================

    df_identifyPrimaryObjects = pd.read_csv(dapi_cp_path)
    df_Phalloidin_segmented = pd.read_csv(phalloidin_cp_path)
    df_YAP_segmented = pd.read_csv(yap_cp_path)

    print("\nLoaded CellProfiler tables:")
    print("DAPI:", df_identifyPrimaryObjects.shape)
    print("Phalloidin:", df_Phalloidin_segmented.shape)
    print("YAP:", df_YAP_segmented.shape)

    # ============================================================
    # 2.6 Prefix CellProfiler columns
    # ============================================================

    df_id = prepare_df(
        df_identifyPrimaryObjects,
        "dapi",
        keys
    )

    df_ph = prepare_df(
        df_Phalloidin_segmented,
        "phalloidin",
        keys
    )

    df_yap = prepare_df(
        df_YAP_segmented,
        "yap",
        keys
    )

    # ============================================================
    # 2.7 Merge pseudotime + DAPI + phalloidin + YAP
    # ============================================================

    before_merge = len(cluster_map)

    SD_clean = (
        cluster_map
        .merge(df_id, on=keys, how="inner")
        .merge(df_ph, on=keys, how="inner")
        .merge(df_yap, on=keys, how="inner")
    )

    after_merge = len(SD_clean)

    print("\nAfter CellProfiler merge:")
    print(f"Before merge: {before_merge}")
    print(f"After merge:  {after_merge}")
    print(f"Removed by inner merge: {before_merge - after_merge}")
    print("Merged SD_clean shape:", SD_clean.shape)

    # ============================================================
    # 2.8 Remove cells with no cytoplasmic area
    # ============================================================

    before_area_filter = len(SD_clean)

    SD_clean = SD_clean[
        SD_clean["yap_AreaShape_Area"] > SD_clean["dapi_AreaShape_Area"]
    ].copy()

    after_area_filter = len(SD_clean)
    removed_area_filter = before_area_filter - after_area_filter

    print("\nCytoplasmic-area filter:")
    print(f"Removed {removed_area_filter} objects where YAP area <= DAPI area")
    print(f"Remaining objects: {after_area_filter}")

    # ============================================================
    # 2.9 Reset index after all filtering / merging
    # ============================================================

    SD_clean = SD_clean.reset_index(drop=True)
    SD_clean["clustering_row_id"] = np.arange(len(SD_clean))

    print("\nFinal database used for clustering:")
    print("SD_clean shape:", SD_clean.shape)

    filter_report = pd.DataFrame([{
        "Feature_Idx": feature_idx,
        "initial_pseudotime_n": n_initial,
        f"after_pseudotime_widths_len_>={min_profile_length}": n_after_length_filter,
        "manual_outliers_listed": len(outliers),
        "after_outlier_filter": n_after_outlier_filter,
        "after_cellprofiler_inner_merge": after_merge,
        "after_yap_area_gt_dapi_area": after_area_filter,
        "final_n_used_for_clustering": len(SD_clean),
    }])

    show(filter_report)

    # ============================================================
    # 2.10 Build fixed-length feature matrix from pseudotime_widths
    # ============================================================

    X_series = SD_clean["pseudotime_widths"].to_list()
    lengths = np.array([len(ts) for ts in X_series], dtype=int)

    target_len = int(np.median(lengths))

    X_shape = np.vstack([
        resample_ts_raw(ts, target_len)
        for ts in X_series
    ])

    # Optional extra feature: original trajectory/profile length
    orig_len = lengths.astype(float).reshape(-1, 1)

    # Final clustering matrix:
    # profile shape + original profile length
    X_features = np.hstack([X_shape, orig_len])

    # ============================================================
    # 2.11 Scale and PCA
    # ============================================================

    X_scaled = StandardScaler().fit_transform(X_features)

    n_pcs_cluster = min(
        n_pcs_cluster_default,
        X_scaled.shape[1],
        X_scaled.shape[0] - 1
    )

    pca = PCA(
        n_components=n_pcs_cluster,
        svd_solver="full"
    )

    X_pca = pca.fit_transform(X_scaled)

    # For plotting: use first 2 PCs from the same PCA
    X_plot = X_pca[:, :2]

    print("\nPCA input/output:")
    print("X_shape shape:", X_shape.shape)
    print("X_features shape:", X_features.shape)
    print("X_pca shape:", X_pca.shape)
    print("Target resampled length:", target_len)
    print("Original profile length included:", True)
    print("Explained variance ratio:", pca.explained_variance_ratio_)
    print("Explained variance sum:", pca.explained_variance_ratio_.sum())

    # ============================================================
    # 2.12 DBSCAN exploration
    # ============================================================

    min_samples = min(min_samples_default, len(SD_clean) - 1)

    nbrs = NearestNeighbors(
        n_neighbors=min_samples
    )

    nbrs.fit(X_pca)

    distances, _ = nbrs.kneighbors(X_pca)
    kdist = np.sort(distances[:, -1])

    plt.figure(figsize=(6, 4))
    plt.plot(kdist)
    plt.title(f"Feature {feature_idx}: DBSCAN k-distance")
    plt.xlabel("Points sorted by distance")
    plt.ylabel(f"Distance to {min_samples}th nearest neighbor")
    plt.tight_layout()
    plt.show()

    dbscan_results = []

    for eps in eps_values:
        labels = DBSCAN(
            eps=eps,
            min_samples=min_samples
        ).fit_predict(X_pca)

        unique, counts = np.unique(labels, return_counts=True)
        sizes = dict(zip(unique, counts))

        mask = labels != -1

        if len(np.unique(labels[mask])) >= 2:
            sil = silhouette_score(X_pca[mask], labels[mask])
        else:
            sil = np.nan

        dbscan_results.append({
            "Feature_Idx": feature_idx,
            "eps": eps,
            "silhouette_non_noise": sil,
            "sizes": sizes,
            "n_noise": int(np.sum(labels == -1)),
            "n_non_noise": int(np.sum(labels != -1)),
        })

        print(f"eps={eps}, silhouette_non_noise={sil}, sizes={sizes}")

    dbscan_summary = pd.DataFrame(dbscan_results)
    show(dbscan_summary)

    # ============================================================
    # 2.13 Agglomerative clustering evaluation
    # ============================================================

    results = []

    for k in ks:
        agg = AgglomerativeClustering(
            n_clusters=k,
            linkage="ward"
        )

        labels = agg.fit_predict(X_pca)

        sil = silhouette_score(X_pca, labels)

        unique, counts = np.unique(labels, return_counts=True)
        sizes = dict(zip(unique, counts))

        colname = f"cluster_agg_k{k}"
        SD_clean[colname] = labels

        results.append({
            "k": k,
            "labels": labels,
            "silhouette": sil,
            "sizes": sizes,
            "colname": colname,
        })

        print(f"k={k}, silhouette={sil:.3f}, sizes={sizes}")

    agg_summary = pd.DataFrame([
        {
            "Feature_Idx": feature_idx,
            "k": r["k"],
            "silhouette": r["silhouette"],
            **{f"cluster_{c}": n for c, n in r["sizes"].items()}
        }
        for r in results
    ]).fillna(0)

    show(agg_summary)

    # ============================================================
    # 2.14 Add PCA coordinates back to SD_clean
    # ============================================================

    SD_clean["PC1"] = X_plot[:, 0]
    SD_clean["PC2"] = X_plot[:, 1]

    # Manuscript-selected final cluster label:
    # 1462 -> k=5
    # 1476 -> k=2
    if feature_idx == 1462:
        SD_clean["cluster_agg_k"] = SD_clean["cluster_agg_k5"]
        SD_clean["cluster_agg_k_source"] = "cluster_agg_k5"

    elif feature_idx == 1476:
        SD_clean["cluster_agg_k"] = SD_clean["cluster_agg_k2"]
        SD_clean["cluster_agg_k_source"] = "cluster_agg_k2"

    else:
        SD_clean["cluster_agg_k"] = np.nan
        SD_clean["cluster_agg_k_source"] = "not_defined"

    # ============================================================
    # 2.15 PCA scatter plots for k=2,3,4,5,6
    # ============================================================

    fig, axes = plt.subplots(
        1,
        len(ks),
        figsize=(6 * len(ks), 8),
        sharex=True,
        sharey=True
    )

    if len(ks) == 1:
        axes = [axes]

    for ax, r in zip(axes, results):
        labels = r["labels"]

        for c in np.unique(labels):
            idx = labels == c
            n_nuclei = np.sum(idx)

            ax.scatter(
                X_plot[idx, 0],
                X_plot[idx, 1],
                s=20,
                alpha=0.7,
                label=f"C{c} (n={n_nuclei})"
            )

        ax.set_title(f"Feature {feature_idx}\nk={r['k']}\nsil={r['silhouette']:.3f}")
        ax.set_xlabel("PC1")
        ax.set_ylabel("PC2")
        ax.legend(title="Clusters")

    plt.tight_layout()
    plt.show()

    # ============================================================
    # 2.16 Table 2-like summary for this surface
    # ============================================================

    table2_like = agg_summary[["Feature_Idx", "k", "silhouette"]].copy()

    table2_wide = (
        table2_like
        .pivot_table(
            index="Feature_Idx",
            columns="k",
            values="silhouette"
        )
        .reset_index()
    )

    table2_wide = table2_wide.rename(columns={
        2: "k=2",
        3: "k=3",
        4: "k=4",
        5: "k=5",
        6: "k=6",
    })

    table2_wide_rounded = table2_wide.copy()

    for col in ["k=2", "k=3", "k=4", "k=5", "k=6"]:
        table2_wide_rounded[col] = table2_wide_rounded[col].round(2)

    show(table2_wide_rounded)

    # ============================================================
    # 2.17 Save outputs for this surface
    # ============================================================

    out_pkl = out_dir / f"feature_{feature_idx}_SD_clean_clustered_with_cellprofiler.pkl"
    out_csv = out_dir / f"feature_{feature_idx}_SD_clean_clustered_with_cellprofiler.csv"

    SD_clean.to_pickle(out_pkl)
    SD_clean.to_csv(out_csv, index=False)

    filter_report.to_csv(
        out_dir / f"feature_{feature_idx}_filter_report.csv",
        index=False
    )

    agg_summary.to_csv(
        out_dir / f"feature_{feature_idx}_agg_summary.csv",
        index=False
    )

    dbscan_summary.to_csv(
        out_dir / f"feature_{feature_idx}_dbscan_summary.csv",
        index=False
    )

    table2_wide_rounded.to_csv(
        out_dir / f"feature_{feature_idx}_table2_like_summary.csv",
        index=False
    )

    print("\nSaved:")
    print(out_pkl)
    print(out_csv)

    print("\nFinal SD_clean shape:", SD_clean.shape)
    print("Final selected cluster column: cluster_agg_k")
    print("PCA coordinate columns: PC1, PC2")

    return {
        "feature_idx": feature_idx,
        "SD_clean": SD_clean,
        "filter_report": filter_report,
        "agg_summary": agg_summary,
        "dbscan_summary": dbscan_summary,
        "table2_wide_rounded": table2_wide_rounded,
        "X_shape": X_shape,
        "X_features": X_features,
        "X_scaled": X_scaled,
        "X_pca": X_pca,
        "X_plot": X_plot,
        "pca": pca,
        "results": results,
        "target_len": target_len,
        "lengths": lengths,
    }


# ============================================================
# 3. RUN BOTH SURFACES
# ============================================================

outputs = {}

for feature_idx in FEATURES:
    outputs[feature_idx] = run_surface_clustering(feature_idx)


# ============================================================
# 4. CONVENIENCE OBJECTS
# ============================================================

SD_1462_clean = outputs[1462]["SD_clean"]
SD_1476_clean = outputs[1476]["SD_clean"]

filter_report_all = pd.concat(
    [outputs[f]["filter_report"] for f in FEATURES],
    ignore_index=True
)

agg_summary_all = pd.concat(
    [outputs[f]["agg_summary"] for f in FEATURES],
    ignore_index=True
)

dbscan_summary_all = pd.concat(
    [outputs[f]["dbscan_summary"] for f in FEATURES],
    ignore_index=True
)

table2_both = (
    agg_summary_all[["Feature_Idx", "k", "silhouette"]]
    .pivot_table(
        index="Feature_Idx",
        columns="k",
        values="silhouette"
    )
    .reset_index()
)

table2_both = table2_both.rename(columns={
    2: "k=2",
    3: "k=3",
    4: "k=4",
    5: "k=5",
    6: "k=6",
})

table2_both_rounded = table2_both.copy()

for col in ["k=2", "k=3", "k=4", "k=5", "k=6"]:
    table2_both_rounded[col] = table2_both_rounded[col].round(2)


# ============================================================
# 5. SAVE COMBINED SUMMARIES
# ============================================================

filter_report_all.to_csv(
    out_dir / "both_surfaces_filter_report.csv",
    index=False
)

agg_summary_all.to_csv(
    out_dir / "both_surfaces_agg_summary.csv",
    index=False
)

dbscan_summary_all.to_csv(
    out_dir / "both_surfaces_dbscan_summary.csv",
    index=False
)

table2_both_rounded.to_csv(
    out_dir / "both_surfaces_table2_like_summary.csv",
    index=False
)


# ============================================================
# 6. DISPLAY FINAL SUMMARIES
# ============================================================

print("\n" + "=" * 100)
print("BOTH SURFACES FILTER REPORT")
print("=" * 100)
show(filter_report_all)

print("\n" + "=" * 100)
print("BOTH SURFACES TABLE 2-LIKE SUMMARY")
print("=" * 100)
show(table2_both_rounded)

print("\nDone.")
print("Available objects:")
print("  outputs")
print("  SD_1462_clean")
print("  SD_1476_clean")
print("  filter_report_all")
print("  agg_summary_all")
print("  dbscan_summary_all")
print("  table2_both_rounded")
```

```python
import numpy as np
import pandas as pd

from pathlib import Path
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.cluster import AgglomerativeClustering
from sklearn.metrics import silhouette_score


# ============================================================
# Reproduce Table 2:
# Silhouette scores for candidate pseudolocation cluster numbers
# ============================================================

ks = [2, 3, 4, 5, 6]
n_pcs_cluster = 5

out_dir = Path("clustering_outputs_both_surfaces")
out_dir.mkdir(exist_ok=True)


# ============================================================
# 1. Get filtered / merged databases from previous block
# ============================================================

# This assumes you already ran the full clustering block for both surfaces.
# These objects should already exist:
#   SD_1462_clean
#   SD_1476_clean
#
# If not, load them from disk.

try:
    SD_1462_table2 = SD_1462_clean.copy()
    SD_1476_table2 = SD_1476_clean.copy()

except NameError:
    SD_1462_table2 = pd.read_pickle(
        out_dir / "feature_1462_SD_clean_clustered_with_cellprofiler.pkl"
    )
    SD_1476_table2 = pd.read_pickle(
        out_dir / "feature_1476_SD_clean_clustered_with_cellprofiler.pkl"
    )


datasets = {
    1462: SD_1462_table2,
    1476: SD_1476_table2,
}


# ============================================================
# 2. Helper functions
# ============================================================

def resample_ts_raw(ts, target_len):
    """
    Resample variable-length pseudotime_widths profile
    into fixed length.
    """
    ts = np.asarray(ts, dtype=float)

    if len(ts) == 1:
        return np.repeat(ts[0], target_len)

    x_old = np.linspace(0, 1, len(ts))
    x_new = np.linspace(0, 1, target_len)

    return np.interp(x_new, x_old, ts)


def build_clustering_matrix(df, include_profile_length=True):
    """
    Build clustering matrix from pseudotime_widths.

    include_profile_length=True:
        profile shape + original profile length

    include_profile_length=False:
        profile shape only
    """
    X_series = df["pseudotime_widths"].to_list()
    lengths = np.array([len(ts) for ts in X_series], dtype=int)

    target_len = int(np.median(lengths))

    X_shape = np.vstack([
        resample_ts_raw(ts, target_len)
        for ts in X_series
    ])

    if include_profile_length:
        orig_len = lengths.astype(float).reshape(-1, 1)
        X_features = np.hstack([X_shape, orig_len])
    else:
        X_features = X_shape

    return X_features, X_shape, lengths, target_len


def compute_silhouette_row(
    df,
    feature_idx,
    include_profile_length=True,
    ks=ks,
    n_pcs_cluster=n_pcs_cluster,
):
    """
    Compute silhouette scores for k=2..6 for one surface
    and one clustering input type.
    """
    if include_profile_length:
        clustering_input = "Profile shape + profile length"
    else:
        clustering_input = "Profile shape only"

    X_features, X_shape, lengths, target_len = build_clustering_matrix(
        df,
        include_profile_length=include_profile_length
    )

    X_scaled = StandardScaler().fit_transform(X_features)

    n_pcs = min(
        n_pcs_cluster,
        X_scaled.shape[1],
        X_scaled.shape[0] - 1
    )

    pca = PCA(
        n_components=n_pcs,
        svd_solver="full"
    )

    X_pca = pca.fit_transform(X_scaled)

    row = {
        "Feature": feature_idx,
        "Clustering input": clustering_input,
        "n_cells": len(df),
        "target_len": target_len,
        "n_features_before_PCA": X_features.shape[1],
        "n_PCs": n_pcs,
        "explained_variance_sum": pca.explained_variance_ratio_.sum(),
    }

    cluster_size_info = {}

    for k in ks:
        agg = AgglomerativeClustering(
            n_clusters=k,
            linkage="ward"
        )

        labels = agg.fit_predict(X_pca)

        sil = silhouette_score(X_pca, labels)

        row[f"k={k}"] = sil

        unique, counts = np.unique(labels, return_counts=True)
        cluster_size_info[f"k={k}_sizes"] = dict(zip(unique, counts))

    return row, cluster_size_info


# ============================================================
# 3. Compute all Table 2 rows
# ============================================================

table2_rows = []
cluster_sizes_rows = []

for feature_idx, df in datasets.items():

    for include_profile_length in [True, False]:

        row, size_info = compute_silhouette_row(
            df=df,
            feature_idx=feature_idx,
            include_profile_length=include_profile_length
        )

        table2_rows.append(row)

        size_row = {
            "Feature": row["Feature"],
            "Clustering input": row["Clustering input"],
            **size_info
        }

        cluster_sizes_rows.append(size_row)


table2_full = pd.DataFrame(table2_rows)
table2_cluster_sizes = pd.DataFrame(cluster_sizes_rows)


# ============================================================
# 4. Manuscript-style Table 2
# ============================================================

table2_manuscript = table2_full[
    [
        "Feature",
        "Clustering input",
        "k=2",
        "k=3",
        "k=4",
        "k=5",
        "k=6",
    ]
].copy()

table2_manuscript_rounded = table2_manuscript.copy()

for col in ["k=2", "k=3", "k=4", "k=5", "k=6"]:
    table2_manuscript_rounded[col] = table2_manuscript_rounded[col].round(2)


# ============================================================
# 5. Display and save
# ============================================================

display(table2_manuscript_rounded)

table2_manuscript.to_csv(
    out_dir / "reproduced_table2_silhouette_scores_full_precision.csv",
    index=False
)

table2_manuscript_rounded.to_csv(
    out_dir / "reproduced_table2_silhouette_scores_rounded.csv",
    index=False
)

table2_full.to_csv(
    out_dir / "reproduced_table2_with_pca_metadata.csv",
    index=False
)

table2_cluster_sizes.to_csv(
    out_dir / "reproduced_table2_cluster_sizes.csv",
    index=False
)

print("Saved:")
print(out_dir / "reproduced_table2_silhouette_scores_full_precision.csv")
print(out_dir / "reproduced_table2_silhouette_scores_rounded.csv")
print(out_dir / "reproduced_table2_with_pca_metadata.csv")
print(out_dir / "reproduced_table2_cluster_sizes.csv")
```

```python
import ast
from pathlib import Path

import numpy as np
import pandas as pd

from scipy.stats import kruskal, spearmanr
from statsmodels.stats.multitest import multipletests


# ============================================================
# 0. Load clustered databases if not already in memory
# ============================================================

out_dir = Path("clustering_outputs_both_surfaces")

try:
    datasets = {
        1462: SD_1462_clean.copy(),
        1476: SD_1476_clean.copy(),
    }
except NameError:
    datasets = {
        1462: pd.read_pickle(out_dir / "feature_1462_SD_clean_clustered_with_cellprofiler.pkl"),
        1476: pd.read_pickle(out_dir / "feature_1476_SD_clean_clustered_with_cellprofiler.pkl"),
    }


probe_out_dir = Path("yap_general_probe")
probe_out_dir.mkdir(exist_ok=True)


# ============================================================
# 1. Helper functions
# ============================================================

def parse_profile(x):
    if isinstance(x, np.ndarray):
        return x.astype(float)

    if isinstance(x, list):
        return np.asarray(x, dtype=float)

    if isinstance(x, tuple):
        return np.asarray(list(x), dtype=float)

    if isinstance(x, str):
        try:
            return np.asarray(ast.literal_eval(x), dtype=float)
        except Exception:
            cleaned = (
                x.replace("[", "")
                 .replace("]", "")
                 .replace(",", " ")
                 .split()
            )
            return np.asarray(cleaned, dtype=float)

    return np.array([], dtype=float)


def mean_profile_value(x):
    arr = parse_profile(x)
    arr = arr[np.isfinite(arr)]
    if len(arr) == 0:
        return np.nan
    return float(np.mean(arr))


def bh_qvalues(pvals):
    pvals = np.asarray(pvals, dtype=float)
    qvals = np.full_like(pvals, np.nan, dtype=float)

    valid = np.isfinite(pvals)

    if valid.sum() > 0:
        qvals[valid] = multipletests(pvals[valid], method="fdr_bh")[1]

    return qvals


def classify_yap_feature(col):
    c = col.lower()

    if "intensity" in c:
        return "Intensity"

    if "texture" in c:
        return "Texture"

    if "granularity" in c:
        return "Granularity"

    if "radialdistribution" in c or "radial_distribution" in c:
        return "RadialDistribution"

    if "areashape" in c or "area_shape" in c:
        return "AreaShape"

    if "neighbors" in c:
        return "Neighbors"

    if "location" in c:
        return "Location"

    return "Other"


def flag_size_sensitive_feature(col):
    c = col.lower()

    size_sensitive_terms = [
        "integratedintensity",
        "totalintensity",
        "area",
        "perimeter",
        "majoraxislength",
        "minoraxislength",
        "feretdiameter",
        "boundingbox",
        "extent",
    ]

    return any(term.lower() in c for term in size_sensitive_terms)


def get_numeric_yap_columns(df):
    yap_cols = [
        c for c in df.columns
        if c.startswith("yap_")
    ]

    numeric_yap_cols = []

    for c in yap_cols:
        s = pd.to_numeric(df[c], errors="coerce")
        if s.notna().sum() >= 10 and s.nunique(dropna=True) > 1:
            numeric_yap_cols.append(c)

    return numeric_yap_cols


def kruskal_feature_test(df, feature_col, group_col, min_group_n=5):
    tmp = df[[feature_col, group_col]].copy()
    tmp[feature_col] = pd.to_numeric(tmp[feature_col], errors="coerce")
    tmp = tmp.dropna()

    groups = []
    group_sizes = {}

    for group_name, group_df in tmp.groupby(group_col):
        vals = group_df[feature_col].dropna().values
        group_sizes[group_name] = len(vals)

        if len(vals) >= min_group_n:
            groups.append(vals)

    if len(groups) < 2:
        return np.nan, np.nan, np.nan, group_sizes, np.nan

    try:
        H, p = kruskal(*groups)
    except Exception:
        return np.nan, np.nan, np.nan, group_sizes, np.nan

    n = sum(len(g) for g in groups)
    k = len(groups)

    if n > k:
        epsilon_sq = (H - k + 1) / (n - k)
        epsilon_sq = max(0, epsilon_sq)
    else:
        epsilon_sq = np.nan

    means = tmp.groupby(group_col)[feature_col].mean()
    mean_range = means.max() - means.min()

    return H, p, epsilon_sq, group_sizes, mean_range


def spearman_feature_test(df, feature_col, x_col, min_n=20):
    tmp = df[[feature_col, x_col]].copy()
    tmp[feature_col] = pd.to_numeric(tmp[feature_col], errors="coerce")
    tmp[x_col] = pd.to_numeric(tmp[x_col], errors="coerce")
    tmp = tmp.dropna()

    if len(tmp) < min_n:
        return np.nan, np.nan, len(tmp)

    try:
        rho, p = spearmanr(tmp[x_col], tmp[feature_col])
    except Exception:
        rho, p = np.nan, np.nan

    return rho, p, len(tmp)


# ============================================================
# 2. Prepare data and inventory YAP features
# ============================================================

prepared = {}
inventory_rows = []

for feature_idx, df in datasets.items():
    df = df.copy()

    if "mean_confinement_width" not in df.columns:
        df["mean_confinement_width"] = df["pseudotime_widths"].apply(mean_profile_value)

    numeric_yap_cols = get_numeric_yap_columns(df)

    prepared[feature_idx] = {
        "df": df,
        "yap_cols": numeric_yap_cols,
    }

    for col in numeric_yap_cols:
        s = pd.to_numeric(df[col], errors="coerce")

        inventory_rows.append({
            "Feature_Idx": feature_idx,
            "YAP_feature": col,
            "family": classify_yap_feature(col),
            "size_sensitive_flag": flag_size_sensitive_feature(col),
            "n_non_null": int(s.notna().sum()),
            "missing_fraction": float(s.isna().mean()),
            "mean": float(s.mean()),
            "std": float(s.std()),
            "cv_abs": float(abs(s.std() / s.mean())) if s.mean() not in [0, np.nan] else np.nan,
            "min": float(s.min()),
            "max": float(s.max()),
        })

yap_inventory = pd.DataFrame(inventory_rows)

print("YAP feature inventory:")
display(
    yap_inventory
    .groupby(["Feature_Idx", "family"])
    .size()
    .reset_index(name="n_features")
)

yap_inventory.to_csv(
    probe_out_dir / "yap_feature_inventory.csv",
    index=False
)


# ============================================================
# 3. Probe YAP differences across clusters
# ============================================================

cluster_results = []

for feature_idx, item in prepared.items():
    df = item["df"]
    yap_cols = item["yap_cols"]

    cluster_cols = [
        c for c in df.columns
        if c.startswith("cluster_agg_k")
        and c != "cluster_agg_k_source"
    ]

    # Ensure final selected cluster is included first
    if "cluster_agg_k" in cluster_cols:
        cluster_cols = ["cluster_agg_k"] + [c for c in cluster_cols if c != "cluster_agg_k"]

    for cluster_col in cluster_cols:
        for yap_col in yap_cols:
            H, p, eps2, sizes, mean_range = kruskal_feature_test(
                df,
                feature_col=yap_col,
                group_col=cluster_col,
                min_group_n=5
            )

            cluster_results.append({
                "Feature_Idx": feature_idx,
                "cluster_column": cluster_col,
                "YAP_feature": yap_col,
                "family": classify_yap_feature(yap_col),
                "size_sensitive_flag": flag_size_sensitive_feature(yap_col),
                "kruskal_H": H,
                "p_value": p,
                "epsilon_squared": eps2,
                "cluster_mean_range": mean_range,
                "group_sizes": sizes,
            })

cluster_results = pd.DataFrame(cluster_results)

cluster_results["q_value"] = (
    cluster_results
    .groupby(["Feature_Idx", "cluster_column"])["p_value"]
    .transform(bh_qvalues)
)

cluster_results["cluster_interest_score"] = (
    -np.log10(cluster_results["q_value"].replace(0, np.nan))
    * cluster_results["epsilon_squared"]
)

cluster_results = cluster_results.sort_values(
    ["Feature_Idx", "cluster_column", "q_value", "epsilon_squared"],
    ascending=[True, True, True, False]
)

cluster_results.to_csv(
    probe_out_dir / "yap_cluster_probe_all_clusterings.csv",
    index=False
)


# ============================================================
# 4. Probe YAP differences across pseudolocation phase
# ============================================================

phase_results = []

for feature_idx, item in prepared.items():
    df = item["df"]
    yap_cols = item["yap_cols"]

    phase_col = "pseudotime_center_bin"

    if phase_col not in df.columns:
        continue

    for yap_col in yap_cols:
        H, p, eps2, sizes, mean_range = kruskal_feature_test(
            df,
            feature_col=yap_col,
            group_col=phase_col,
            min_group_n=5
        )

        phase_results.append({
            "Feature_Idx": feature_idx,
            "YAP_feature": yap_col,
            "family": classify_yap_feature(yap_col),
            "size_sensitive_flag": flag_size_sensitive_feature(yap_col),
            "kruskal_H": H,
            "p_value": p,
            "epsilon_squared": eps2,
            "phase_mean_range": mean_range,
            "phase_bin_sizes": sizes,
        })

phase_results = pd.DataFrame(phase_results)

phase_results["q_value"] = (
    phase_results
    .groupby("Feature_Idx")["p_value"]
    .transform(bh_qvalues)
)

phase_results["phase_interest_score"] = (
    -np.log10(phase_results["q_value"].replace(0, np.nan))
    * phase_results["epsilon_squared"]
)

phase_results = phase_results.sort_values(
    ["Feature_Idx", "q_value", "epsilon_squared"],
    ascending=[True, True, False]
)

phase_results.to_csv(
    probe_out_dir / "yap_phase_probe.csv",
    index=False
)


# ============================================================
# 5. Probe YAP association with mean confinement width
# ============================================================

width_results = []

for feature_idx, item in prepared.items():
    df = item["df"]
    yap_cols = item["yap_cols"]

    for yap_col in yap_cols:
        rho, p, n = spearman_feature_test(
            df,
            feature_col=yap_col,
            x_col="mean_confinement_width",
            min_n=20
        )

        width_results.append({
            "Feature_Idx": feature_idx,
            "YAP_feature": yap_col,
            "family": classify_yap_feature(yap_col),
            "size_sensitive_flag": flag_size_sensitive_feature(yap_col),
            "spearman_rho": rho,
            "abs_spearman_rho": abs(rho) if pd.notna(rho) else np.nan,
            "p_value": p,
            "n": n,
        })

width_results = pd.DataFrame(width_results)

width_results["q_value"] = (
    width_results
    .groupby("Feature_Idx")["p_value"]
    .transform(bh_qvalues)
)

width_results["width_interest_score"] = (
    -np.log10(width_results["q_value"].replace(0, np.nan))
    * width_results["abs_spearman_rho"]
)

width_results = width_results.sort_values(
    ["Feature_Idx", "q_value", "abs_spearman_rho"],
    ascending=[True, True, False]
)

width_results.to_csv(
    probe_out_dir / "yap_confinement_width_probe.csv",
    index=False
)


# ============================================================
# 6. Combine into one ranked feature-interest table
# ============================================================

# Best cluster result per YAP feature per surface
best_cluster = (
    cluster_results
    .sort_values(
        ["Feature_Idx", "YAP_feature", "q_value", "epsilon_squared"],
        ascending=[True, True, True, False]
    )
    .groupby(["Feature_Idx", "YAP_feature"], as_index=False)
    .first()
    [[
        "Feature_Idx",
        "YAP_feature",
        "cluster_column",
        "q_value",
        "epsilon_squared",
        "cluster_mean_range",
        "cluster_interest_score",
    ]]
    .rename(columns={
        "cluster_column": "best_cluster_column",
        "q_value": "best_cluster_q",
        "epsilon_squared": "best_cluster_epsilon_squared",
        "cluster_mean_range": "best_cluster_mean_range",
    })
)

phase_small = phase_results[
    [
        "Feature_Idx",
        "YAP_feature",
        "q_value",
        "epsilon_squared",
        "phase_mean_range",
        "phase_interest_score",
    ]
].rename(columns={
    "q_value": "phase_q",
    "epsilon_squared": "phase_epsilon_squared",
})

width_small = width_results[
    [
        "Feature_Idx",
        "YAP_feature",
        "q_value",
        "spearman_rho",
        "abs_spearman_rho",
        "width_interest_score",
    ]
].rename(columns={
    "q_value": "width_q",
})

interesting_yap = (
    yap_inventory[
        [
            "Feature_Idx",
            "YAP_feature",
            "family",
            "size_sensitive_flag",
            "n_non_null",
            "missing_fraction",
            "std",
            "cv_abs",
        ]
    ]
    .merge(best_cluster, on=["Feature_Idx", "YAP_feature"], how="left")
    .merge(phase_small, on=["Feature_Idx", "YAP_feature"], how="left")
    .merge(width_small, on=["Feature_Idx", "YAP_feature"], how="left")
)

for col in [
    "cluster_interest_score",
    "phase_interest_score",
    "width_interest_score",
]:
    interesting_yap[col] = interesting_yap[col].replace([np.inf, -np.inf], np.nan).fillna(0)

interesting_yap["combined_interest_score"] = (
    interesting_yap["cluster_interest_score"]
    + interesting_yap["phase_interest_score"]
    + interesting_yap["width_interest_score"]
)

interesting_yap = interesting_yap.sort_values(
    ["Feature_Idx", "combined_interest_score"],
    ascending=[True, False]
)

interesting_yap.to_csv(
    probe_out_dir / "yap_interesting_features_ranked.csv",
    index=False
)


# ============================================================
# 7. Display useful first-look tables
# ============================================================

print("\nTop YAP features overall per surface:")
display(
    interesting_yap
    .groupby("Feature_Idx")
    .head(25)
    [[
        "Feature_Idx",
        "YAP_feature",
        "family",
        "size_sensitive_flag",
        "combined_interest_score",
        "best_cluster_column",
        "best_cluster_q",
        "best_cluster_epsilon_squared",
        "phase_q",
        "phase_epsilon_squared",
        "width_q",
        "spearman_rho",
    ]]
)

print("\nTop YAP features for final selected manuscript clusters only:")
final_cluster_top = (
    cluster_results[cluster_results["cluster_column"] == "cluster_agg_k"]
    .sort_values(
        ["Feature_Idx", "q_value", "epsilon_squared"],
        ascending=[True, True, False]
    )
)

display(
    final_cluster_top
    .groupby("Feature_Idx")
    .head(25)
    [[
        "Feature_Idx",
        "YAP_feature",
        "family",
        "size_sensitive_flag",
        "q_value",
        "epsilon_squared",
        "cluster_mean_range",
        "group_sizes",
    ]]
)

print("\nTop YAP features associated with pseudolocation phase:")
display(
    phase_results
    .groupby("Feature_Idx")
    .head(25)
    [[
        "Feature_Idx",
        "YAP_feature",
        "family",
        "size_sensitive_flag",
        "q_value",
        "epsilon_squared",
        "phase_mean_range",
    ]]
)

print("\nTop YAP features associated with mean confinement width:")
display(
    width_results
    .groupby("Feature_Idx")
    .head(25)
    [[
        "Feature_Idx",
        "YAP_feature",
        "family",
        "size_sensitive_flag",
        "q_value",
        "spearman_rho",
        "abs_spearman_rho",
        "n",
    ]]
)


# ============================================================
# 8. Save compact top tables
# ============================================================

interesting_yap.groupby("Feature_Idx").head(50).to_csv(
    probe_out_dir / "top50_interesting_yap_features_per_surface.csv",
    index=False
)

final_cluster_top.groupby("Feature_Idx").head(50).to_csv(
    probe_out_dir / "top50_yap_features_final_clusters.csv",
    index=False
)

phase_results.groupby("Feature_Idx").head(50).to_csv(
    probe_out_dir / "top50_yap_features_phase.csv",
    index=False
)

width_results.groupby("Feature_Idx").head(50).to_csv(
    probe_out_dir / "top50_yap_features_width.csv",
    index=False
)

print("\nSaved all probe outputs to:")
print(probe_out_dir.resolve())
```
